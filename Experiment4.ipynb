```python
#program to find first and last colors
colors=input("Enter colors seperated by comma: ")
list=colors.split(",")
print("first color:",list[0])
print("last color:",list[-1])

```

```python
#extension of file
filename=input("Enter filename: ")
extension=filename.split(".")[-1]
print("extension:",extension)
```

```python
#program which accepts a sequence of comma seperated numbers and generate list,sum of numbers
numbers=input("enter numbers seperated by commas: ")
numbers=list(map(int,numbers.split(",")))


print("Listed numbers:",numbers)
print("Sum of numbers:",sum(numbers))
```
