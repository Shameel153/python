```python
#get a string from a given string where all occurences of its first char have been changed to $,except the first char itself
s=input("Enter the string: ")
a=s[0]
b=s[1:]
b=b.replace(a,"$")
print(a+b)
```

```python
#to change to a new string where the first and last chars have been exchanged
n= input("enter the string: ")
print(n[5] + n[1:5] + n[0])
```

```python
#create a single string seperated with space from two strings by swapping the character at position 1
s1 = input("enter first string: ")
s2 = input("enter second string: ")
r1 = s2[0]+s1[1:]
r2 = s1[0]+s2[1:]
print(r1,r2)
```
